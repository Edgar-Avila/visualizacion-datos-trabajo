```python
from pathlib import Path
import pandas as pd
import duckdb

# Ruta del proyecto
PROYECTO = Path.cwd().parent

# Carpeta con los datos procesados
SILVER = PROYECTO / "preprocesamiento" / "silver"

# Verificar archivos
print("Carpeta silver:", SILVER.resolve())
print("Existe:", SILVER.exists())

# Abrir el resumen mensual
resumen = pd.read_parquet(SILVER / "resumen_mensual.parquet")

display(resumen.head())
```

```python
# Leer los datos agregados
agregado = pd.read_parquet(SILVER / "consumo_agregado.parquet")

# Mostrar las primeras 5 filas
display(agregado.head())

# Ver dimensiones: filas y columnas
print("Dimensiones:", agregado.shape)

# Ver nombres de columnas
print("Columnas:", agregado.columns.tolist())
```

```python
# Ruta del archivo
archivo_clientes = (SILVER / "consumo_clientes.parquet").as_posix()

# Consultar las primeras 10 filas
muestra_clientes = duckdb.sql(f"""
    SELECT *
    FROM read_parquet('{archivo_clientes}')
    LIMIT 10
""").df()

display(muestra_clientes)
```
